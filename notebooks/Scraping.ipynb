```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time

# 配置 Chrome 无头模式
options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")
driver = webdriver.Chrome(options=options)

# 打开页面
driver.get("https://www.flightsfrom.com/LOS")
time.sleep(5)  # 等待页面加载完 JS

# 查找目的地列表
destinations = driver.find_elements(By.CSS_SELECTOR, ".destinations li a")  # 根据网页结构调整选择器

for dest in destinations:
    name = dest.text
    link = dest.get_attribute("href")
    print(name, link)

driver.quit()
```
